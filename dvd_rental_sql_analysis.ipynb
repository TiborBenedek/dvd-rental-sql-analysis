# DVD Rental Database – SQL Analysis

This notebook explores the PostgreSQL DVD Rental sample database using SQL queries.

The goal of this analysis is to explore revenue patterns, customer behavior, and product performance using SQL aggregation and relational joins.

SQL dialect: PostgreSQL

## Dataset Overview

The DVD Rental database is a sample relational dataset commonly used for practicing SQL queries and database analysis.

It simulates a DVD rental business with customers, film inventory, rental transactions, and payments.

Main tables used in this analysis:

- `payment` – transaction payments from rentals
- `customer` – customer information
- `rental` – rental transactions
- `inventory` – film copies available in stores
- `film` – movie metadata
- `category` – film categories
- `film_category` – relationship between films and categories
- `store` – store locations
- `address`, `city`, `country` – geographic data

The analysis focuses on revenue aggregation and customer behavior.

## 1. Total Revenue

This query calculates the total revenue generated from all rental payments.

````python
```sql
SELECT
    SUM(amount) AS total_revenue
FROM payment;
```
````

## 2. Revenue per Month

This query aggregates revenue by month to observe how income changes over time.

````python
```sql
SELECT
    date_trunc('month', payment_date) AS payment_month,
    SUM(amount) AS revenue
FROM payment
GROUP BY payment_month
ORDER BY payment_month;
```
````

## 3. Top Customers by Revenue

This query identifies customers who generated the highest revenue through rentals.

````python
```sql
SELECT
    customer.customer_id,
    first_name,
    last_name,
    SUM(amount) AS total_spend
FROM payment
JOIN customer
    ON customer.customer_id = payment.customer_id
GROUP BY
    customer.customer_id,
    first_name,
    last_name
ORDER BY total_spend DESC
LIMIT 10;
```
````

## 4. Customers with the Most Rentals

This query counts the number of rentals made by each customer.

````python
```sql
SELECT
    rental.customer_id,
    first_name,
    last_name,
    COUNT(rental.rental_id) AS total_rentals
FROM rental
JOIN customer
    ON customer.customer_id = rental.customer_id
GROUP BY
    rental.customer_id,
    first_name,
    last_name
ORDER BY total_rentals DESC;
```
````

## 5. Most Rented Films

This query identifies films with the highest number of rentals.

````python
```sql
SELECT
    title,
    COUNT(rental_id) AS total_rentals
FROM rental
JOIN inventory
    ON inventory.inventory_id = rental.inventory_id
JOIN film
    ON film.film_id = inventory.film_id
GROUP BY title
ORDER BY total_rentals DESC
LIMIT 10;
```
````

## 6. Revenue by Film Category

This query calculates total revenue generated by each film category.

````python
```sql
SELECT
    category.name AS category,
    SUM(amount) AS revenue
FROM category
JOIN film_category
    ON category.category_id = film_category.category_id
JOIN inventory
    ON inventory.film_id = film_category.film_id
JOIN rental
    ON rental.inventory_id = inventory.inventory_id
JOIN payment
    ON payment.rental_id = rental.rental_id
GROUP BY category.name
ORDER BY revenue DESC;
```
````

## 7. Revenue by Store

This query calculates revenue generated by each store location.

````python
```sql
SELECT
    store.store_id,
    SUM(amount) AS revenue
FROM store
JOIN customer
    ON customer.store_id = store.store_id
JOIN payment
    ON payment.customer_id = customer.customer_id
GROUP BY store.store_id
ORDER BY revenue DESC;
```
````

## 8. Revenue by Country

This query analyzes geographic revenue distribution based on customer locations.

````python
```sql
SELECT
    country.country,
    SUM(amount) AS revenue
FROM payment
JOIN customer
    ON payment.customer_id = customer.customer_id
JOIN address
    ON address.address_id = customer.address_id
JOIN city
    ON city.city_id = address.city_id
JOIN country
    ON country.country_id = city.country_id
GROUP BY country.country
ORDER BY revenue DESC;
```
````

## Key Insights

- Revenue is generated through a large number of relatively small rental transactions.
- A small group of customers contributes significantly to total revenue.
- Some films and categories are rented much more frequently than others.
- Revenue varies across stores and geographic locations.
- Understanding customer behavior can help improve marketing and inventory decisions.
